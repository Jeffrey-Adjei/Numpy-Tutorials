```python
# define the sine function

sample_rate = 50000 # Hz

frequency = 2 # Hz

time = 10 # seconds

def sine_wave_function(frequency, sample_rate, duration):
    x = np.linspace(0, duration, sample_rate * duration)
    frequencies = x

    y = np.sin((2 * np.pi) * frequencies)
    return x , y
```

```python
x, y = sine_wave_function(2, 80000, 5)
plt.plot(x, y)
plt.grid(True)
plt.show
```
