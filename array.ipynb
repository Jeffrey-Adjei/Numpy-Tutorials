```python
import numpy as np
import math
import matplotlib.pyplot as plt
```

```python
# arange
arr = np.arange(10)
print(arr)
```

```python
# step-wise
arr1 = np.arange(0, 10, 2)
print(arr1)
```

```python
arr2 = np.zeros(10)
print(arr2)
```

```python
arr3 = np.ones(10)
print(arr3)
```

```python
arr4 = np.linspace(0, 100, num=10)
print(arr4)
```

```python
# creating a sine & cosine waves using np.linspace

# generate 25 evenspaced points between 0 & 10 for the x-axis
x = np.linspace(0, 10, 25)

# generate the sine and cosine values
y1 = np.sin(x)
y2 = np.cos(x)

# plot the grid 
plt.plot(x, y1, color="r", label="Line 1 - sin wave")
plt.plot(x, y2, color="b" ,label="Line 2 - cos wave")

plt.xlabel("")
plt.ylabel("")
plt.title("Sinusiodal Waves")

plt.grid(True)
plt.legend("")
plt.show
```
